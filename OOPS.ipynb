# __init__() method

* It is a method which automatically gets called when obj of a class is created
* It is used to initialise values to the attributes

```python
class Student:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def display(self):
        print("Name:",self.name)
        print("Age:",self.age)

s1=Student("Tom",20)
s2=Student("Mickey",25)
s1.display()
s2.display()
```

# Inheritance

It is the property where one class inherites attributes & methods of parent class.
* Parent class/ base class
* Child class / derived class

```python
#single inheritence
class Dog:   #parent class
    name="Tommy"
    def bark(self):
        print("Doggy barks")

class Puppy(Dog):  #child class
    name="Tubby"
    def cry(self):
        print("Puppy cries")
p1=Puppy()   #by creating obj of child class we can access the details of parent class
p1.bark()
p1.cry()
```

```python
#Multilevel inheritence
class Grandfather:
    def job(self):
        print("Grandfather was a policeman")
class Father(Grandfather):
    def job1(self):
        print("Father is a Businessman")
class Child(Father):
    def job2(self):
        print("Child is a student")
c=Child()
c.job()
c.job1()
c.job2()

f=Father()
f.job()
f.job2()  #error  -- as father cannot access child details
```

```python
#multiple inheritance
class Father:
    def work(self):
        print("Father is working")
class Mother:
    def cook(self):
        print("Mother is cooking")
class Child(Father,Mother):
    def study(self):
        print("Child is studying")
c=Child()
c.work()
c.cook()
```

```python
#hierarchical inheritance
class Father:
    def work(self):
        print("Working as Teacher")
class Son(Father):
    def study(self):
        print("Studying data science")
class Daughter(Father):
    def dance(self):
        print("Learning Bharatnatyam")
s=Son()
s.work()

d=Daughter()
d.work()
```

## Polymorphism

Methods with same name behaves in different way in different classes

```python
class Cow:
    def sound(self):
        print("Cow can moo")
class Dog:
    def sound(self):
        print("Dog can bark")
class Horse:
    def sound(self):
        print("Horse can neigh")
c=Cow()
c.sound()
d=Dog()
d.sound()
h=Horse()
h.sound()
```

```python
class Rectangle:
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def area(self):
        return self.l*self.b
class Circle:
    def __init__(self,r):
        self.r=r
    def area(self):
        return 3.14*self.r**2
rect=Rectangle(10,2)
print(rect.area())

c=Circle(7)
print(c.area())
```

# Encapsulation

* It means wrapping up attributes & methods into a single unit
* It restricts the access of attributes & methods from outside the class to avoid accidental modification.
* Encapsulation can be implemented using access modifiers like public, private & protected

```python
#public modifier-- is by default the access modifier of all classes
#Anyone can access & modify the public attributes & method
```

```python
#public
class Student:
    def __init__(self,name,age):
        self.name=name  #public 
        self.age=age   #public
    def display(self):  #public
        print("Name:",self.name)
        print("Age:",self.age)
s=Student("Jeevitha",20)   
print(s.name)
s.name="Siya"   #update public attribute name to siya
print(s.name)
```

```python
#private
class Student:
    def __init__(self,name,age):
        self.__name=name    #private
        self.__age=age   
    def __display(self):    #private
        print("Name:",self.__name)
        print("Age:",self.__age)
    def getDetails(self):   #to access private details use public method present in same class
        self.__display()
s=Student("Jeevitha",20)   
#print(s.__name)   #error
#print(s.__display())
s.getDetails()
```

```python
#protected -- only protected details are accessed by derived class
#In python, protected members are truely not protected. _ just denotes that it is a protected member.
```

```python
#protected
class Teacher:
    def __init__(self, subject):
        self.subject=subject
    def _display(self):  #protected method
        print("Subject:",self.subject)
class Student(Teacher):
    def __init__(self,subject,marks):
        Teacher.__init__(self,subject)
        self.marks=marks

    def details(self):
        self._display()
        print("Marks:",self.marks)
s=Student("Maths",20)
s.details()
```

# Abstraction

* It is the ability to hide implementation details of object from user & only exposing its essential details
*  It tell what an object does instead of how it does.

# Abstract class & Abstract method

* Abstract class contains abstract method which nothing but an empty method without any implementation.
* Child class contains implementation of abstract method

```python
from abc import ABC, abstractmethod
```

```python
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass
class Car(Vehicle):
    def start(self):
        print("Engine can start when we press the button")
```

```python
c=Car()
c.start()
```

```python
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
class Rectangle(Shape):
    def __init__(self,l,b):
        self.l=l
        self.b=b
    def area(self):
        return self.l*self.b

r=Rectangle(10,2)
r.area()
```
