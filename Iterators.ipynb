```python
str1="Hello world"
type(str1)
```

```python
str_iter=iter(str1)   #iter()-- which converts a data structure into iterator
```

```python
type(str_iter)
```

```python
next(str_iter)   #next() is used to iterate through iterators.   It throws stop iteration error we reach end of the string 
```

```python
#Customer iterator --- which prints a given count of numbers
class count:
    def __init__(self, limit):
       self.limit=limit
       self.current=1

    def __iter__(self):
        return self

    def __next__(self): 
        if self.current<=self.limit:
            value=self.current
            self.current=self.current+1
            return value
        else:
            raise StopIteration

counter=count(7)
#for num in counter:
#print(num)
count_iter=iter(counter)
print(next(count_iter))
print(next(count_iter))
print(next(count_iter))
print(next(count_iter))
print(next(count_iter))
print(next(count_iter))
print(next(count_iter))
print(next(count_iter))
```
