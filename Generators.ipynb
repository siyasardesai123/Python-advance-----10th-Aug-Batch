* Generators are special kind of functions which generates values one at a time
* They are created just like normal functions but when we use yield keyword, they become generator function

```python
def greet():   #normal userdefined function
    print("Hello")
    print("Hi")
```

```python
def greet():    #generator function
    yield "Hello"
    yield "Hi"
```

* They are called as lazy functions
* They are more memory effiecient & generate any number of values

```python
for i in greet():  #generater function are iterable objects
    print(i)
```

```python
#Write a generator function to generate squares of first 5 numbers
```

```python
def square(n):
    for i in range(1,n+1):
        yield i**2
```

```python
for i in square(5):
    print(i)
```

```python
#write a generator function for given list, to generate capital words for each
list1=['hi','hello','bye']
```

```python
def capitals(list1):
    for i in list1:
        yield i.upper()
```

```python
for i in capitals(list1):
    print(i)
```
