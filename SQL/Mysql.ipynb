```python
## making connections to database server
connection=pymysql.connect(host = 'localhost',user='root',password='abc123', database='college6' )
```

```python
# Create a cursor object to execute SQL queries
cursor = connection.cursor()
```

## Create Table

```python
create_table_query = """
CREATE TABLE IF NOT EXISTS users (
    id INT PRIMARY KEY AUTO_INCREMENT,
    name VARCHAR(50),
    age INT,
    city VARCHAR(50)
)
"""
cursor.execute(create_table_query)

# SQL Query to insert a record
insert_query = "INSERT INTO users (name, age, city) VALUES (%s, %s, %s)"
values = ('Alice', 30, 'New York')

# Execute the query
cursor.execute(insert_query, values)

# Commit the transaction to save changes to the database
connection.commit()

print("Record inserted successfully")
```

```python
db_tables=pd.read_sql_query('show tables',connection)
print(db_tables)
```

## Show Details from Table

```python
query = "select * from users"
output = pd.read_sql(query,connection)
print(output.shape)
output
```

## Inserting Multiple Values

```python
# Insert multiple records
values = [
    ('Bob', 25, 'Los Angeles'),
    ('Charlie', 35, 'Chicago'),
    ('Diana', 28, 'San Francisco')
]
cursor.executemany(insert_query, values)
connection.commit()
```

```python
# SQL Query to select all records
select_query = "SELECT * FROM users"

# Execute the query
cursor.execute(select_query)

# Fetch all the rows from the result
rows = cursor.fetchall()

# Print the results
for row in rows:
    print(row)
```

```python
query = "select * from users"
output = pd.read_sql(query,connection)
print(output.shape)
output
```

## Using Where Clause

```python
# SQL Query to select specific records (e.g., users older than 30)
select_query = "SELECT * FROM users Where AGE > %s"
cursor.execute(select_query, (30,))

# Fetch and print the filtered rows
rows = cursor.fetchall()
for row in rows:
    print(row)
```

## Updating Records

```python
# SQL Query to update a record
update_query = "UPDATE users SET city = %s WHERE name = %s"
values = ('Las Vegas', 'Bob')

# Execute the update query
cursor.execute(update_query, values)

# Commit the transaction
connection.commit()

print("Record updated successfully")
```

```python
query = "select * from users"
output = pd.read_sql(query,connection)
print(output.shape)
output
```

## Deleting Record

```python
# SQL Query to delete a record
delete_query = "DELETE FROM users WHERE name = %s"
value = ('Charlie',)

# Execute the delete query
cursor.execute(delete_query, value)

# Commit the transaction
connection.commit()

print("Record deleted successfully")
```

```python
query = "select * from users"
output = pd.read_sql(query,connection)
print(output.shape)
output
```
