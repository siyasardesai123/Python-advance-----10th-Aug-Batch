* Decorators are the function which extends functionality of one function by wrapping it inside another function without changing its code.
* We define a decorator by putting @ anotation above it

```python
def greet_extend(func):
    def wrapper():
        print("Hello")
        func()
        print("Bye")
    return wrapper
```

```python
@greet_extend
def greet():
    print("Good afternoon")

greet()
```

```python
@div_extend
def div(x,y):
    return x/y
```

```python
div(2,4)
```

```python
def div_extend(func):
    def wrapper(x,y):
        if x<y:
            x,y=y,x
        return func(x,y)
    return wrapper
```

```python
def capital_extend(func):
    def wrapper(text):
        if text.startswith('s') or text.startswith('S'):
            print("String Starts with s") 
            return func(text)
        else:
            print("String does not start with s") 
            return func(text)     
    return wrapper
```

```python
# create a function to square a num. Extend its behaviour by checking if the number is odd or even
```

```python
def square_extend(func):
    def wrapper(num):
        if num%2==0:
            print(num,"is even")
            func(num)
        else:
            print(num,"is odd")
            func(num)
    return wrapper
```

```python
@square_extend
def square(num):
    print(num**2)
```

```python
square(7)
```

# Static method

* These are methods in class without self keyword
* Because of that  these method cannot be used access & modify any attributes of the class
* They are defined by using @staticmethod annotation above it

```python
class Calculator:
    @staticmethod
    def add(a,b):
        return a+b
```

```python
Calculator.add(5,2)
```

```python
c=Calculator()
c.add(4,5)
```

```python
class oddeven:
    @staticmethod
    def iseven(num):
        if num%2==0:
            return "Even"
        else:
            return "Odd"
```

```python
oddeven.iseven(5)
```

# Double underscore methods

* Also called as dunder methods as we have 2 underscores before & after their name
* These methods are automatically called when the operations on objects are performed.
* Operator overloading can be performed using these methods

```python
class student:
    def __init__(self,name):
        self.name=name
    def __str__(self):
        return f'Name of student is:{self.name}'
```

```python
s=student("Geeta")
print(s)
```

```python
class Numbers:
    def __init__(self,value):  #initialises value to attribute
        self.value=value

    def __add__(self,other):    #adds values of 2 objs
        return self.value+other.value

    def __mul__(self,other):    #multiplies values of 2 objs
        return self.value*other.value
    
    def __lt__(self,other):    #compares values of 2 objs
        return self.value<other.value
x=int(input("Enter a num:"))
n1=Numbers(x)
n2=Numbers(15)
n3= Numbers(7)
print(n1*n2)
print(n1<n2)
```

```python
class student:
    def __init__(self,m1,m2,m3):
        self.m1=m1
        self.m2=m2
        self.m3=m3

    def __add__(self,other):
        return self.m1+other.m1
```

```python
s1=student(10,20,30)
s2=student(30,40,50)
s3=student(70,90,100)
s1+s2
```
