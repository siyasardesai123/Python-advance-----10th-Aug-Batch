MongoDB, a NoSQL database, is designed to handle large amounts of data using a flexible, JSON-like format known as BSON (Binary JSON).

MongoDB supports these operations via the PyMongo library in Python

## First establish a connection to the MongoDB server. The default port for MongoDB is 27017

```python
from pymongo import MongoClient

# Create a MongoClient instance
client = MongoClient("mongodb://localhost:27017/")

# Connect to a specific database (create it if it doesn't exist)
db = client["my_database4"]

# Connect to a specific collection (similar to a table in relational databases)
collection = db["my_collection"]
```

# Create (Insert) Documents

MongoDB collections consist of documents, which are similar to rows in relational databases. Documents are stored in BSON format (which is like JSON).

```python
# Insert a single document (record)
document = {"name": "Alice", "age": 30, "city": "New York"}
collection.insert_one(document)
```

```python
# Insert multiple documents
documents = [
    {"name": "Bob", "age": 25, "city": "Los Angeles"},
    {"name": "Charlie", "age": 35, "city": "Chicago"},
    {"name": "Diana", "age": 28, "city": "San Francisco"}
]
collection.insert_many(documents)
```

# Read (Query) Documents

You can retrieve documents from MongoDB using the find or find_one methods.

```python
# Find a single document
result = collection.find_one({"city": "New York"})
print(result)
```

```python
# Find all documents in the collection
results = collection.find()

# Iterate through the results
for document in results:
    print(document)
```

```python
# Find documents where age is greater than 30
results = collection.find({"age": {"$gt": 30}})

for document in results:
    print(document)
```

# Update Documents

```python
# Update a single document (e.g., change Alice's age to 32)
collection.update_one({"name": "Charlie"}, {"$set": {"city": "Las Vegas"}})
```

```python
# Update multiple documents (e.g., increase age by 1 for all people over 25)
collection.update_many({"age": {"$gt": 25}}, {"$inc": {"age": 1}})
```

```python
# Find all documents in the collection
results = collection.find()

# Iterate through the results
for document in results:
    print(document)
```

# Delete Documents

```python
# Delete a single document (e.g., delete Bob's record)
collection.delete_one({"name": "Bob"})
```

```python
# Delete multiple documents (e.g., delete all people aged over 30)
collection.delete_many({"age": {"$gt": 30}})
```

```python
# Find all documents in the collection
results = collection.find()

# Iterate through the results
for document in results:
    print(document)
```

# Close the Connection

Always close the MongoDB connection when you're done working with it.

```python
client.close()
```
