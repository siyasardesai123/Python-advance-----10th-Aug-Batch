Exception handling is the process of handling errors so that code can execute smoothly without any hault.
* try block-- will hold the code which might have error & scans the code for errors
* except block -- which print message when error occurs
* finally block -- is the block which always executes if error is there or not there.

```python
a=5
try:
    a+'7'   #type error
except TypeError:
    print("Type error has occured!")
a+2
```

```python
#Name error -- it comes when we try to use variable which is not defined
try:
    a=8
    b=9
    print(a+b)
    print(m)   #m is not defined
except NameError:
    print("Name error!")
print(a-b)
```
