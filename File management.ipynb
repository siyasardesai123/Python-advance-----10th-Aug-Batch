* File management means performing different operations on file such as reading a file, opening a file, writing to a file, updating a file.
* open() is an imp function using which all the operations can be performed on file
* open() has 2 parameters: filename, modes

```python
f1=open('filename','modes')
```

## modes in open()

1) r --- reading a file
2) w-- creates a file if not exists & opens & writes a content to the file
3) a(append)-- creates a file if not exist & opens a file for updating
4) x(create) --creates a file if not exist & throws error as FileExist error if file alredy present
5) r+ -- reads & writes a file
6) rb-- reading a binary file
7) wb-- writing to binary file

```python
#x-- create
f1=open("test.txt","x")
f1.close()
```

```python
f1=open('test.txt','w')
f1.write("hello this is my test file")
f1.close()
```

```python
f1=open('test.txt','w')  #write mode overwrite the earlier content in file
f1.write("Welcome to test")   
f1.close()
```

```python
f1=open('test.txt','a')  #append mode adds the new content at the end of file
f1.write("Welcome to test again")   
f1.close()
```

```python
f1=open('test.txt','r') #read mode is the default mode
content=f1.read()   #reads all the contents of file
print(content)
f1.close()
```

```python
f1=open('test2.txt','x') 
f1.write("hello \n hi \n thanks \n")
f1.close()
```

```python
f1=open('test2.txt','r') #read mode is the default mode
content=f1.readline()  #reads first line from the file 
print(content)
f1.close()
```

```python
f1=open('test2.txt','r') #read mode is the default mode
content=f1.readlines() #reads all the lines in list format
print(content)
f1.close()
```

```python
f1=open('test2.txt','r+')  #opens a file for reading & write to a file
print(f1.read())
f1.write("hello \n hi \n thanks \n")
f1.close()
```

## with statement ensures that whichever file opened will closed automatically

```python
with open("hello.py","w") as f1:
    f1.write("1+2")
```

```python
with open("hello.py") as f1:
    print(f1.read())
```
