```python
from tkinter import *
root= Tk()

root.title("Multiply")
root.geometry("400x200")

result_label=Label(root)
entry1=Entry(root)

def multiply():
    x=entry1.get()
    result=int(x)*4
    result_label["text"]=result
    
btn = Button(root, text="Multiply", command=multiply)
entry1.pack()
btn.pack()
result_label.pack()
root.mainloop()
```

```python
from tkinter import *
from tkinter.ttk import Combobox
from tkinter import messagebox

window=Tk()
window.title('Hello Python')
window.geometry("400x300")

data=["one", "two", "three", "four"]
cb=Combobox(window, values=data)
cb.place(x=60, y=150)

e1=Entry(window)
e1.pack()

r1=Radiobutton(window, text="male", value=1)
r2=Radiobutton(window, text="female", value=2)
r1.place(x=100,y=50)
r2.place(x=180, y=50)


C1 = Checkbutton(window, text = "Cricket",onvalue=1,offvalue=0)
C2 = Checkbutton(window, text = "Tennis",onvalue=1,offvalue=0)
C1.place(x=100, y=100)
C2.place(x=180, y=100)

def Click():
    x=e1.get()  #gets value from entry i.e text box
    messagebox.showinfo("Hi",x)
b1=Button(window,text="Click here",command=Click)
b1.pack()


window.mainloop()
```

```python
from tkinter import *
from tkinter import messagebox
login=Tk()
login.title("Login")
login.geometry("300x300")
login.config(bg="pink")

label=Label(login,text="Username",fg="blue",bg="pink",font=("arial",14,"bold"))
label.place(x=150,y=100,anchor=CENTER)

txtbox=Entry(login)
txtbox.place(x=150,y=150,anchor=CENTER)

def log():
    user=txtbox.get()
    if user=='siya' or user=='rupali' or user=='apurva':
        login.destroy()
        new=Tk()
        new.title("Login Success")
        new.geometry("300x300")
        n1=Label(new,text="Welcome!!",font=("arial",14,"bold"))
        n1.place(x=100,y=100)
        new.mainloop()
    else:
        messagebox.showinfo("Error","Not an authorised user!")

btn=Button(login,text="LOGIN",font=("arial",15,"bold"),command=log)
btn.place(x=150,y=200,anchor=CENTER)
login.mainloop()
```

```python
from tkinter import *
window=Tk()
window.title("Clicker")
window.geometry("200x200")

label1=Label(window)

sum=0  #global variable
def click_here():
    global sum
    sum=sum+1
    label1["text"]=sum

def reset_here():
    global sum
    sum=0
    label1["text"]=sum
button1=Button(window,text="Click",command=click_here)
button2=Button(window,text="Reset",command=reset_here)
label1.pack()
button1.pack()
button2.pack()

window.mainloop()
```
